### LSTM Sequence Preparation and Chronological Purging

The retail purchase-history dataset generated 30,750 sequences for next-purchase interval prediction. Each sequence contains five purchase-event timesteps and 34 input features, with padding handled through an explicit Boolean mask.

To prevent temporal leakage, chronological partitioning was performed using event timestamps and next-purchase target dates. Training and validation samples whose target outcomes crossed their respective partition boundaries were excluded from the final partitions.

A total of 4,133 sequences were purged: 2,626 training candidates and 1,507 validation candidates. The final dataset contains 26,617 sequences: 20,318 for training, 3,177 for validation, and 3,122 for testing. This represents an overall retention rate of approximately 86.56%.

The exclusions are intentional and do not represent deleted raw transactions. The purging strategy prioritizes the integrity of chronological model evaluation over retaining every generated sequence. The final partition counts and temporal validation checks are recorded in the pipeline audit for reproducibility.

## **``Step 3 preparing baseline LSTM model:``**

````python
import tensorflow as tf
from tensorflow.keras import Model
from tensorflow.keras.layers import Input, LSTM, Dropout, Dense
from tensorflow.keras.optimizers import Adam

# Reproducibility
tf.keras.utils.set_random_seed(42)

# Read dimensions from the validated training data
sequence_length = X_train_scaled.shape[1]
n_features = X_train_scaled.shape[2]

# Sequence input
sequence_input = Input(
    shape=(sequence_length, n_features),
    name="sequence_input"
)

# Explicit Boolean mask: True = real event, False = padding
event_mask = Input(
    shape=(sequence_length,),
    dtype="bool",
    name="event_mask"
)

# Baseline LSTM architecture
x = LSTM(units=64,name="lstm_layer")(sequence_input, mask=event_mask)
x = Dropout(rate=0.20,name="dropout_layer")(x)
x = Dense(units=32,activation="relu",name="dense_layer")(x)

# Regression output: predicted days until next purchase
output = Dense(units=1,activation="linear", name="predicted_days")(x)

baseline_lstm = Model(inputs=[sequence_input, event_mask],outputs=output,name="baseline_lstm_regressor")
baseline_lstm.compile(optimizer=Adam(learning_rate=0.001),loss="mse",metrics=["mae"])

baseline_lstm.summary()
````

### **``Step 3.1 baseline sample prediction:``**

````python
# Check the model output shape
sample_predictions = baseline_lstm(
    {
        "sequence_input": X_train_scaled[:4],
        "event_mask": X_train_mask[:4],
    },
    training=False,
)

print("Model name:", baseline_lstm.name)
print("Input shape:", X_train_scaled.shape[1:])
print("Expected output shape:", (4, 1))
print("Actual output shape:", sample_predictions.shape)
print("All predictions finite:",
      bool(tf.reduce_all(tf.math.is_finite(sample_predictions)).numpy()))
````

### **``Step 3.2 Baseline LSTM training``**

````python
import numpy as np
import tensorflow as tf

early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_mae",
    patience=5,
    restore_best_weights=True,
    mode="min",
    verbose=1,
)

history = baseline_lstm.fit(
    x={
        "sequence_input": X_train_scaled,
        "event_mask": X_train_mask,
    },
    y=y_train,
    validation_data=(
        {
            "sequence_input": X_val_scaled,
            "event_mask": X_val_mask,
        },
        y_val,
    ),
    epochs=50,
    batch_size=64,
    callbacks=[early_stopping],
    verbose=1,
)
````

````python
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Baseline prediction: training-target median
median_prediction = float(np.median(y_train))

median_val_predictions = np.full(
    shape=len(y_val),
    fill_value=median_prediction,
)

median_val_mae = mean_absolute_error(
    y_val,
    median_val_predictions,
)

median_val_rmse = np.sqrt(
    mean_squared_error(y_val, median_val_predictions)
)

print(f"Training median prediction: {median_prediction:.2f} days")
print(f"Median baseline validation MAE: {median_val_mae:.2f} days")
print(f"Median baseline validation RMSE: {median_val_rmse:.2f} days")
````

````python
val_predictions = baseline_lstm.predict(
    {
        "sequence_input": X_val_scaled,
        "event_mask": X_val_mask,
    },
    verbose=0,
).reshape(-1)

lstm_val_mae = mean_absolute_error(y_val, val_predictions)

lstm_val_rmse = np.sqrt(
    mean_squared_error(y_val, val_predictions)
)

print(f"LSTM validation MAE: {lstm_val_mae:.2f} days")
print(f"LSTM validation RMSE: {lstm_val_rmse:.2f} days")

print("\nComparison")
print(f"Median baseline MAE: {median_val_mae:.2f} days")
print(f"LSTM MAE:            {lstm_val_mae:.2f} days")
````

````python
# Experiment 2: Same architecture, Huber loss
tf.keras.utils.set_random_seed(42)

huber_lstm = tf.keras.models.clone_model(baseline_lstm)

huber_lstm.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss=tf.keras.losses.Huber(delta=1.0),
    metrics=["mae"],
)

huber_early_stopping = tf.keras.callbacks.EarlyStopping(
    monitor="val_mae",
    patience=5,
    restore_best_weights=True,
    mode="min",
    verbose=1,
)

huber_history = huber_lstm.fit(
    x={
        "sequence_input": X_train_scaled,
        "event_mask": X_train_mask,
    },
    y=y_train,
    validation_data=(
        {
            "sequence_input": X_val_scaled,
            "event_mask": X_val_mask,
        },
        y_val,
    ),
    epochs=50,
    batch_size=64,
    callbacks=[huber_early_stopping],
    verbose=1,
)
````

````python
import numpy as np
import pandas as pd
import tensorflow as tf

from sklearn.metrics import mean_absolute_error, mean_squared_error

# Use only the already-scaled sequences
X_train = X_train_scaled
X_val = X_val_scaled

# Keep the original chronological split and masks
train_inputs = {
    "sequence_input": X_train,
    "event_mask": X_train_mask,
}

val_inputs = {
    "sequence_input": X_val,
    "event_mask": X_val_mask,
}

delta_values = [1.0, 5.0, 10.0, 20.0]

delta_results = []
delta_models = {}
delta_histories = {}

for delta in delta_values:
    print(f"\nTraining Huber LSTM with delta={delta}")

    # Same seed and architecture for a controlled comparison
    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(42)

    model = tf.keras.models.clone_model(baseline_lstm)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.001
        ),
        loss=tf.keras.losses.Huber(delta=delta),
        metrics=["mae"],
    )

    early_stopping = tf.keras.callbacks.EarlyStopping(
        monitor="val_mae",
        patience=5,
        restore_best_weights=True,
        mode="min",
        verbose=0,
    )

    history_delta = model.fit(
        x=train_inputs,
        y=y_train,
        validation_data=(val_inputs, y_val),
        epochs=50,
        batch_size=64,
        callbacks=[early_stopping],
        verbose=0,
    )

    # Evaluate using the restored best weights
    predictions = model.predict(val_inputs, verbose=0).reshape(-1)

    val_mae = mean_absolute_error(y_val, predictions)
    val_rmse = np.sqrt(mean_squared_error(y_val, predictions))

    best_epoch = (
        int(np.argmin(history_delta.history["val_mae"])) + 1
    )

    delta_results.append({
        "Huber_Delta": delta,
        "Best_Epoch": best_epoch,
        "Validation_MAE": val_mae,
        "Validation_RMSE": val_rmse,
    })

    delta_models[delta] = model
    delta_histories[delta] = history_delta

delta_results_df = (
    pd.DataFrame(delta_results)
    .sort_values("Validation_MAE")
    .reset_index(drop=True)
)

display(delta_results_df.round(3))

print(f"Median baseline validation MAE: {median_val_mae:.3f} days")
````

````python
learning_rates = [0.0001, 0.0003, 0.001]

learning_rate_results = []
learning_rate_models = {}
learning_rate_histories = {}

for lr in learning_rates:
    print(f"\nTraining with learning rate={lr}")

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(42)

    model = tf.keras.models.clone_model(baseline_lstm)

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=lr
        ),
        loss=tf.keras.losses.Huber(delta=10.0),
        metrics=["mae"],
    )

    early_stopping = tf.keras.callbacks.EarlyStopping(
        monitor="val_mae",
        patience=7,
        restore_best_weights=True,
        mode="min",
        verbose=0,
    )

    history_lr = model.fit(
        x=train_inputs,
        y=y_train,
        validation_data=(val_inputs, y_val),
        epochs=60,
        batch_size=64,
        callbacks=[early_stopping],
        verbose=0,
    )

    predictions = model.predict(
        val_inputs, verbose=0
    ).reshape(-1)

    val_mae = mean_absolute_error(y_val, predictions)
    val_rmse = np.sqrt(
        mean_squared_error(y_val, predictions)
    )

    best_epoch = int(
        np.argmin(history_lr.history["val_mae"]) + 1
    )

    learning_rate_results.append({
        "Learning_Rate": lr,
        "Best_Epoch": best_epoch,
        "Epochs_Completed": len(history_lr.history["loss"]),
        "Validation_MAE": val_mae,
        "Validation_RMSE": val_rmse,
    })

    learning_rate_models[lr] = model
    learning_rate_histories[lr] = history_lr

learning_rate_results_df = (
    pd.DataFrame(learning_rate_results)
    .sort_values("Validation_MAE")
    .reset_index(drop=True)
)

display(learning_rate_results_df.round(3))

print(f"Median baseline MAE: {median_val_mae:.3f} days")
print("Current champion: delta=10, learning rate=0.001")
print("Current champion MAE: 17.256 days")
````

````python
unit_values = [32, 64, 128]

unit_results = []
unit_models = {}
unit_histories = {}

for units in unit_values:
    print(f"\nTraining LSTM with {units} units")

    tf.keras.backend.clear_session()
    tf.keras.utils.set_random_seed(42)

    # Build a model with the selected number of LSTM units
    sequence_input = tf.keras.Input(
        shape=(X_train_scaled.shape[1], X_train_scaled.shape[2]),
        name="sequence_input",
    )

    event_mask = tf.keras.Input(
        shape=(X_train_mask.shape[1],),
        dtype="bool",
        name="event_mask",
    )

    x = tf.keras.layers.LSTM(
        units=units,
        name="lstm_layer",
    )(sequence_input, mask=event_mask)

    x = tf.keras.layers.Dropout(0.20)(x)
    x = tf.keras.layers.Dense(32, activation="relu")(x)
    output = tf.keras.layers.Dense(1, activation="linear")(x)

    model = tf.keras.Model(
        inputs=[sequence_input, event_mask],
        outputs=output,
    )

    model.compile(
        optimizer=tf.keras.optimizers.Adam(
            learning_rate=0.0001
        ),
        loss=tf.keras.losses.Huber(delta=10.0),
        metrics=["mae"],
    )

    early_stopping = tf.keras.callbacks.EarlyStopping(
        monitor="val_mae",
        patience=7,
        restore_best_weights=True,
        mode="min",
        verbose=0,
    )

    history_units = model.fit(
        x=train_inputs,
        y=y_train,
        validation_data=(val_inputs, y_val),
        epochs=60,
        batch_size=64,
        callbacks=[early_stopping],
        verbose=0,
    )

    predictions = model.predict(
        val_inputs, verbose=0
    ).reshape(-1)

    val_mae = mean_absolute_error(y_val, predictions)
    val_rmse = np.sqrt(mean_squared_error(y_val, predictions))

    unit_results.append({
        "LSTM_Units": units,
        "Best_Epoch": int(
            np.argmin(history_units.history["val_mae"]) + 1
        ),
        "Validation_MAE": val_mae,
        "Validation_RMSE": val_rmse,
        "Parameters": model.count_params(),
    })

    unit_models[units] = model
    unit_histories[units] = history_units

unit_results_df = (
    pd.DataFrame(unit_results)
    .sort_values("Validation_MAE")
    .reset_index(drop=True)
)

display(unit_results_df.round(3))

print("Current champion MAE: 17.148 days")
print("Median baseline MAE: 18.192 days")
````
